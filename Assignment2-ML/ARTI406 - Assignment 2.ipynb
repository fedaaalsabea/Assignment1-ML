### 3.2 Apply Missing Value Strategy 2: Median Imputation

```python
df_missing.head(10)
```

```python
df_imputed = df.copy()

cols_to_drop = [
    'Historical Wards 2003-2015',
    'Zip Codes',
    'Community Areas',
    'Census Tracts',
    'Wards'
]

df_imputed = df_imputed.drop(columns=cols_to_drop)


df_imputed['License #'] = df_imputed['License #'].fillna(df_imputed['License #'].median())
df_imputed['Zip'] = df_imputed['Zip'].fillna(df_imputed['Zip'].median())
df_imputed['Latitude'] = df_imputed['Latitude'].fillna(df_imputed['Latitude'].median())
df_imputed['Longitude'] = df_imputed['Longitude'].fillna(df_imputed['Longitude'].median())


df_imputed['AKA Name'] = df_imputed['AKA Name'].fillna(df_imputed['AKA Name'].mode()[0])
df_imputed['Facility Type'] = df_imputed['Facility Type'].fillna(df_imputed['Facility Type'].mode()[0])
df_imputed['Risk'] = df_imputed['Risk'].fillna(df_imputed['Risk'].mode()[0])
df_imputed['City'] = df_imputed['City'].fillna(df_imputed['City'].mode()[0])
df_imputed['State'] = df_imputed['State'].fillna(df_imputed['State'].mode()[0])
df_imputed['Inspection Type'] = df_imputed['Inspection Type'].fillna(df_imputed['Inspection Type'].mode()[0])
df_imputed['Violations'] = df_imputed['Violations'].fillna(
    df_imputed['Violations'].mode()[0]
)

df_imputed['Location'] = df_imputed['Location'].fillna(
    df_imputed['Location'].mode()[0]
)


df_imputed.isnull().sum()
```

```python
df_imputed.head(10)
```

Median Imputation was selected to handle missing values.

The median is less sensitive to outliers than the mean and provides a more robust estimate for numerical features.

Using this strategy allowed the dataset to retain all records while replacing missing values with representative values.

Columns with extremely large numbers of missing values (Historical Wards 2003-2015, Zip Codes, Community Areas, Census Tracts, and Wards) were removed because most of their values were missing.

## Task 3 Detect and handle outliers using IQR.
### Handling Outliers


```python
df_imputed.describe()
```

```python
numeric_cols = ['License #', 'Zip', 'Latitude', 'Longitude']

for col in numeric_cols:
    plt.figure(figsize=(5,3))
    sns.boxplot(x=df_imputed[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
for col in numeric_cols:

    Q1 = df_imputed[col].quantile(0.25)
    Q3 = df_imputed[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_imputed[
        (df_imputed[col] < lower_bound) |
        (df_imputed[col] > upper_bound)
    ]

    print(f"{col}: {len(outliers)} outliers")
    
```

### Detect Outliers using IQR

```python
df_outliers = df_imputed.copy()

for col in numeric_cols:

    Q1 = df_outliers[col].quantile(0.25)
    Q3 = df_outliers[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_outliers[col] = df_outliers[col].clip(lower=lower_bound,
                                             upper=upper_bound)

 
```

```python
for col in numeric_cols:

    Q1 = df_outliers[col].quantile(0.25)
    Q3 = df_outliers[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_outliers[
        (df_outliers[col] < lower_bound) |
        (df_outliers[col] > upper_bound)
    ]

    print(f"{col}: {len(outliers)}")
```

### Remove Outliers

```python
df_no_outliers = df_imputed.copy()

for col in numeric_cols:
    Q1 = df_no_outliers[col].quantile(0.25)
    Q3 = df_no_outliers[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_no_outliers = df_no_outliers[
        (df_no_outliers[col] >= lower) &
        (df_no_outliers[col] <= upper)
    ]

print(df_imputed.shape)
print(df_no_outliers.shape)
```

### Capping Outliers

```python
 

df_outliers = df_imputed.copy()

numeric_cols = ['License #', 'Zip', 'Latitude', 'Longitude']

for col in numeric_cols:
    Q1 = df_outliers[col].quantile(0.25)
    Q3 = df_outliers[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_outliers[col] = df_outliers[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )
```

```python
for col in numeric_cols:
    Q1 = df_outliers[col].quantile(0.25)
    Q3 = df_outliers[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df_outliers[
        (df_outliers[col] < lower_bound) |
        (df_outliers[col] > upper_bound)
    ]

    print(f"{col}: {len(outliers)} outliers")
```

 

Outliers were detected using the IQR method. The initial analysis found outliers in License #, Zip, Latitude, and Longitude. To preserve all records, the outliers were handled using IQR-based capping, where values outside the lower and upper bounds were replaced with the corresponding boundary values. After applying capping, no outliers remained in the selected numerical columns.

## Task 5 Apply PCA and interpret explained variance.

### Check Correlation Before Applying PCA

```python
plt.figure(figsize=(6,4))

sns.heatmap(
    df_zscore[numeric_cols].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap Before PCA")
plt.show()
```

The heatmap shows the correlation among the numerical features (License #, Zip, Latitude, and Longitude).

### Data Reduction – Principal Component Analysis (PCA)

```python
from sklearn.decomposition import PCA

X = df_zscore[numeric_cols]

pca = PCA()

principal_components = pca.fit_transform(X)

print("Explained Variance Ratio:")
print(pca.explained_variance_ratio_)
```

The explained variance ratio shows how much variance is captured by each principal component.

- PC1 explains approximately 39.48% of the variance.
- PC2 explains approximately 25.03% of the variance.
- PC3 explains approximately 20.78% of the variance.
- PC4 explains approximately 14.72% of the variance.

The first two principal components explain approximately 64.50% of the total variance. The first three components explain approximately 85.26% of the total variance.

This indicates that PCA can reduce dimensionality while preserving most of the important information in the dataset.

```python
from sklearn.decomposition import PCA

X = df_zscore[numeric_cols]

pca = PCA(n_components=2)

pca_result = pca.fit_transform(X)

# Plot PCA

plt.figure(figsize=(6,4))

plt.scatter(
    pca_result[:,0],
    pca_result[:,1],
    alpha=0.5
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Projection")
plt.grid(True)

plt.show()
```

Each point represents one observation in the dataset after PCA transformation.

The horizontal axis represents Principal Component 1 (PC1), and the vertical axis represents Principal Component 2 (PC2).

These components are combinations of the original numerical features and capture the maximum variance in the data.

The plot provides a two-dimensional representation of the dataset after dimensionality reduction.

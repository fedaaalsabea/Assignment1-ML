# Assignment 1: Exploratory Data Analysis (EDA)

EDA is the first and most important step in any Machine Learning project.
Before building models, we must understand:

- What does the data represent?
- Are there missing values?
- Are there outliers?
- What patterns exist?
- Which variables influence others?

If we do not understand the data, we cannot build a good model.

### Why EDA is Important

In real-world machine learning projects:

    1- 70–80% of the time is spent on understanding and cleaning data
    2- Only 20–30% is spent building models

**EDA helps us:**

- Detect errors
- Identify trends
- Discover relationships
- Make business decisions

### Data after cleaning
### Missing Values

```python
 
df = df.drop(columns=['Historical Wards 2003-2015','Zip Codes','Community Areas','Census Tracts','Wards'], errors='ignore')

 
df = df.drop(columns=['Latitude', 'Longitude', 'Location'], errors='ignore')
text_columns = ['AKA Name', 'Facility Type', 'Risk', 'City', 'State', 'Inspection Type']
for col in text_columns:
    df[col] = df[col].fillna('Unknown')
df['Violations'] = df['Violations'].fillna('No Violations')
df['License #'] = df['License #'].fillna(df['License #'].median())
df['Zip'] = df['Zip'].fillna(df['Zip'].mode()[0])
df['Inspection Date'] = pd.to_datetime(df['Inspection Date'], errors='coerce')
df = df.dropna(subset=['Inspection Date'])
df['Year'] = df['Inspection Date'].dt.year
df.isnull().sum()
```

### duplicate rows


```python
df = df.drop_duplicates()
df.duplicated()[df.duplicated()==True]
```

### Univariate Analysis

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Risk'], bins=20,color="#c386af")
plt.title("Distribution of Risk")
plt.show()
```

- The chart shows the distribution of inspection risk levels.
- Most inspections are classified as "Risk 1 (High)".
- "Risk 2 (Medium)" appears less frequently.
- "Risk 3 (Low)" has the lowest number of inspections.

### Distribution of Inspection IDs

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Inspection ID'], bins=20,color="#66a4bb")
plt.title("Distribution of Inspection ID")
plt.show()
```

## Bivariate Analysis

### Inspections by Result

```python
results_count = df.groupby('Results')['Inspection ID'].count().sort_values(ascending=False)

plt.figure(figsize=(8,5))
results_count.plot(color="#bb7766",kind='bar')
plt.title("Total Inspections by Result")
plt.ylabel("Number of Inspections")
plt.xlabel("Results")
plt.show()

results_count
```

### Inspection ID by risk


```python
risk_count = df.groupby('Risk')['Inspection ID'].count().sort_values(ascending=False)
plt.figure(figsize=(10,5))
risk_count.plot(color="#667abb",kind='bar')
plt.title("Inspections by Risk Level")
plt.ylabel("Number of Inspections")
plt.show()

risk_count
```

### Number of Inspections by Facility Type

```python
product_revenue = df.groupby('Facility Type')['Inspection ID'].count().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
product_revenue.plot(color="#ca96e0",kind='bar')
plt.title("Top 10 Facility Types by Inspections")
plt.ylabel("Number of Inspections")
plt.show()

product_revenue
```

### Number of Inspections by Result

```python
results_count = df.groupby('Results')['Inspection ID'].count().sort_values(ascending=False)

results_count.head(10)
```

### Number of Inspections per Year

```python
year_count = df.groupby(df['Inspection Date'].dt.year)['Inspection ID'].count()

plt.figure(figsize=(8,5))
year_count.plot(color="#7b5f87",kind='line')
plt.title("Number of Inspections per Year")
plt.ylabel("Count")
plt.show()
```

### Correlation Analysis Between Inspection ID and Zip 

```python
plt.figure(figsize=(6,4))
sns.heatmap(df[['Inspection ID','Zip']].corr(),annot=True,color="#886a94")
plt.title("Correlation Matrix")
plt.show()
```

- Correlation close to 1 = strong positive relationship
- Close to 0 = weak relationship
- The correlation value between Inspection ID and Zip Code is approximately 0, indicating no significant relationship between the two variables.

## Time-Based Analysis of Inspection Results
### Yearly Inspection Results Distribution

```python
df['Year'] = df['Inspection Date'].dt.year

results_year = df.groupby(['Year', 'Results']).size().unstack()

plt.figure(figsize=(8,5))
results_year.plot(linewidth=2)

plt.title("Inspection Results Over Years")
plt.xlabel("Year")
plt.ylabel("Count")
plt.grid(alpha=0.3)
plt.show()
```

- The plot shows the trend of inspection results over the years. The number of "Pass" results is consistently higher compared to other categories, indicating that most inspections meet the required standards. Some categories such as "Fail" and "No Entry" appear less frequently, suggesting fewer violations or access issues over time.

### Exploring the Relationship Between Year and License 


```python
plt.figure(figsize=(8,5))
sns.scatterplot(x=df['Year'], y=df['License #'], color="#37356a", alpha=0.3)
plt.title("License Number Distribution Over Years")
plt.xlabel("Year")
plt.ylabel("License #")
plt.show()
```

### Analysis of Inspection Results Distribution

```python
results_count = df['Results'].value_counts().head(4)

plt.figure(figsize=(6,6))
plt.pie(results_count,labels=results_count.index,autopct='%1.1f%%',colors=["#d6e59f", "#c9caed", "#e5a8b6", "#daab72"])
plt.title("Inspection Results Distribution")
plt.show()
```
